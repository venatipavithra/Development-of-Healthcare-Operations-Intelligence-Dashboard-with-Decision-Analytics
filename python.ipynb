### Problem 1: Reverse a String

**Problem Statement:** Write a Python function that takes a string as input and returns the string reversed. Do not use built-in reverse functions/methods.

```python
def reverse_string(s):
    reversed_s = ''
    for char in s:
        reversed_s = char + reversed_s
    return reversed_s

# Example usage:
print(f"'hello' reversed is: {reverse_string('hello')}")
print(f"'Python' reversed is: {reverse_string('Python')}")
```

### Problem 2: Find the Largest Element in a List

**Problem Statement:** Write a Python function to find the largest element in a given list of numbers. Do not use the `max()` function.

```python
def find_largest_element(numbers):
    if not numbers:
        return None # Handle empty list case
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest

# Example usage:
print(f"Largest in [1, 5, 2, 8, 3] is: {find_largest_element([1, 5, 2, 8, 3])}")
print(f"Largest in [-10, -2, -5] is: {find_largest_element([-10, -2, -5])}")
print(f"Largest in [] is: {find_largest_element([])}")
```

### Problem 3: Count Vowels in a String

**Problem Statement:** Write a Python function that takes a string as input and returns the number of vowels (a, e, i, o, u) in it. The function should be case-insensitive.

```python
def count_vowels(s):
    vowels = "aeiou"
    count = 0
    for char in s:
        if char.lower() in vowels:
            count += 1
    return count

# Example usage:
print(f"Number of vowels in 'Hello World': {count_vowels('Hello World')}")
print(f"Number of vowels in 'Python Programming': {count_vowels('Python Programming')}")
print(f"Number of vowels in 'rhythm': {count_vowels('rhythm')}")
```

### Problem 4: Factorial of a Number

**Problem Statement:** Write a Python function to calculate the factorial of a non-negative integer. The factorial of a number `n` is the product of all positive integers less than or equal to `n` (e.g., 5! = 5 * 4 * 3 * 2 * 1 = 120). Handle edge cases like 0! and negative numbers.

```python
def factorial(n):
    if not isinstance(n, int) or n < 0:
        return "Input must be a non-negative integer"
    if n == 0:
        return 1
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Example usage:
print(f"Factorial of 5 is: {factorial(5)}")
print(f"Factorial of 0 is: {factorial(0)}")
print(f"Factorial of 7 is: {factorial(7)}")
print(f"Factorial of -3 is: {factorial(-3)}")
```

### Problem 5: Fibonacci Sequence

**Problem Statement:** Write a Python function to generate the first `n` numbers in the Fibonacci sequence. The Fibonacci sequence starts with 0 and 1, and each subsequent number is the sum of the two preceding ones (e.g., 0, 1, 1, 2, 3, 5, 8...). Handle cases for `n=0` and `n=1`.

```python
def fibonacci_sequence(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    else:
        sequence = [0, 1]
        while len(sequence) < n:
            next_num = sequence[-1] + sequence[-2]
            sequence.append(next_num)
        return sequence

# Example usage:
print(f"Fibonacci sequence for n=7: {fibonacci_sequence(7)}")
print(f"Fibonacci sequence for n=1: {fibonacci_sequence(1)}")
print(f"Fibonacci sequence for n=0: {fibonacci_sequence(0)}")
print(f"Fibonacci sequence for n=10: {fibonacci_sequence(10)}")
```

### Problem 6: Check for Palindrome

**Problem Statement:** Write a Python function that checks if a given string is a palindrome. A palindrome reads the same forwards and backwards (e.g., 'madam', 'racecar'). Ignore cases and non-alphanumeric characters.

```python
import re

def is_palindrome(s):
    # Convert to lowercase and remove non-alphanumeric characters
    s = re.sub(r'[^a-zA-Z0-9]', '', s).lower()
    return s == s[::-1]

# Example usage:
print(f"Is 'madam' a palindrome? {is_palindrome('madam')}")
print(f"Is 'A man, a plan, a canal: Panama' a palindrome? {is_palindrome('A man, a plan, a canal: Panama')}")
print(f"Is 'hello' a palindrome? {is_palindrome('hello')}")
```

### Problem 7: Find Common Elements in Two Lists

**Problem Statement:** Write a Python function that takes two lists as input and returns a new list containing only the elements that are common to both lists. The order of elements in the output list does not matter. Ensure no duplicates in the result.

```python
def find_common_elements(list1, list2):
    return list(set(list1) & set(list2))

# Example usage:
list_a = [1, 2, 3, 4, 5]
list_b = [4, 5, 6, 7, 8]
print(f"Common elements in {list_a} and {list_b}: {find_common_elements(list_a, list_b)}")

list_c = ['apple', 'banana', 'orange']
list_d = ['banana', 'grape', 'apple']
print(f"Common elements in {list_c} and {list_d}: {find_common_elements(list_c, list_d)}")

list_e = [1, 1, 2, 3]
list_f = [2, 3, 3, 4]
print(f"Common elements in {list_e} and {list_f}: {find_common_elements(list_e, list_f)}")
```

### Problem 8: Check for Anagrams

**Problem Statement:** Write a Python function that checks if two strings are anagrams of each other. Anagrams are words or phrases formed by rearranging the letters of another (e.g., 'listen' and 'silent'). Ignore cases and spaces.

```python
def are_anagrams(s1, s2):
    # Remove spaces and convert to lowercase
    s1 = s1.replace(' ', '').lower()
    s2 = s2.replace(' ', '').lower()

    # Check if sorted strings are equal
    return sorted(s1) == sorted(s2)

# Example usage:
print(f"Are 'listen' and 'silent' anagrams? {are_anagrams('listen', 'silent')}")
print(f"Are 'Debit Card' and 'Bad Credit' anagrams? {are_anagrams('Debit Card', 'Bad Credit')}")
print(f"Are 'hello' and 'world' anagrams? {are_anagrams('hello', 'world')}")
```

### Problem 9: Check if a Number is Prime

**Problem Statement:** Write a Python function that takes an integer as input and returns `True` if it's a prime number, `False` otherwise. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself. Handle edge cases.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage:
print(f"Is 7 a prime number? {is_prime(7)}")
print(f"Is 10 a prime number? {is_prime(10)}")
print(f"Is 2 a prime number? {is_prime(2)}")
print(f"Is 1 a prime number? {is_prime(1)}")
print(f"Is 29 a prime number? {is_prime(29)}")
```

### Problem 10: Count Word Frequency in a Sentence

**Problem Statement:** Write a Python function that takes a sentence (string) as input and returns a dictionary where keys are the words in the sentence and values are their respective frequencies (how many times they appear). The function should be case-insensitive and ignore punctuation.

```python
import re

def word_frequency(sentence):
    # Convert to lowercase and remove punctuation
    sentence = re.sub(r'[^a-zA-Z\s]', '', sentence).lower()
    words = sentence.split()

    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1
    return frequency

# Example usage:
text1 = "Hello world, this is a test. Hello again."
print(f"Word frequency for '{text1}': {word_frequency(text1)}")

text2 = "The quick brown fox jumps over the lazy dog. The dog barks."
print(f"Word frequency for '{text2}': {word_frequency(text2)}")
```
