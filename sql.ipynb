## SQL Interview Questions

### Question 1: Find all employees who earn more than their manager.

**Problem Statement:** You have an `Employees` table with columns `employee_id`, `employee_name`, `salary`, and `manager_id`. Write a SQL query to find the names of employees who earn more than their direct manager.

**Solution:**
```sql
SELECT
    e1.employee_name
FROM
    Employees e1
JOIN
    Employees e2 ON e1.manager_id = e2.employee_id
WHERE
    e1.salary > e2.salary;
```

### Question 2: Second Highest Salary

**Problem Statement:** Write a SQL query to find the second highest salary from the `Employees` table. If there is no second highest salary, the query should return `NULL`.

**Solution (using LIMIT/OFFSET for MySQL/PostgreSQL):**
```sql
SELECT DISTINCT
    salary
FROM
    Employees
ORDER BY
    salary DESC
LIMIT 1 OFFSET 1;
```

**Solution (using CTE and DENSE_RANK for SQL Server/Oracle):**
```sql
WITH RankedSalaries AS (
    SELECT
        salary,
        DENSE_RANK() OVER (ORDER BY salary DESC) as rank_num
    FROM
        Employees
)
SELECT
    CASE WHEN COUNT(*) < 2 THEN NULL ELSE MAX(salary) END
FROM
    RankedSalaries
WHERE
    rank_num = 2;
```

### Question 3: Department with the Highest Average Salary

**Problem Statement:** Given an `Employees` table (columns: `employee_id`, `employee_name`, `salary`, `department_id`) and a `Departments` table (columns: `department_id`, `department_name`), find the department with the highest average salary.

**Solution:**
```sql
SELECT
    d.department_name
FROM
    Employees e
JOIN
    Departments d ON e.department_id = d.department_id
GROUP BY
    d.department_name
ORDER BY
    AVG(e.salary) DESC
LIMIT 1;
```

### Question 4: Employees who have never made an order

**Problem Statement:** You have an `Employees` table (columns: `employee_id`, `employee_name`) and an `Orders` table (columns: `order_id`, `employee_id`, `order_date`). Find all employees who have never placed an order.

**Solution (using LEFT JOIN and WHERE IS NULL):**
```sql
SELECT
    e.employee_name
FROM
    Employees e
LEFT JOIN
    Orders o ON e.employee_id = o.employee_id
WHERE
    o.order_id IS NULL;
```

**Solution (using NOT IN):**
```sql
SELECT
    e.employee_name
FROM
    Employees e
WHERE
    e.employee_id NOT IN (SELECT DISTINCT employee_id FROM Orders);
```

### Question 5: Count of orders per customer

**Problem Statement:** You have `Customers` table (columns: `customer_id`, `customer_name`) and `Orders` table (columns: `order_id`, `customer_id`, `order_date`). Write a SQL query to display the customer name and the total number of orders placed by each customer.

**Solution:**
```sql
SELECT
    c.customer_name,
    COUNT(o.order_id) AS total_orders
FROM
    Customers c
LEFT JOIN
    Orders o ON c.customer_id = o.customer_id
GROUP BY
    c.customer_name
ORDER BY
    total_orders DESC;
```

### Question 6: Find Duplicate Emails

**Problem Statement:** Write a SQL query to find all duplicate emails in a `Person` table. The table has columns `id` and `email`.

**Solution:**
```sql
SELECT
    email
FROM
    Person
GROUP BY
    email
HAVING
    COUNT(email) > 1;
```

### Question 7: Customers Who Never Ordered

**Problem Statement:** Write a SQL query to find all customers who never ordered anything. You have a `Customers` table (id, name) and an `Orders` table (id, customerId).

**Solution:**
```sql
SELECT
    c.name
FROM
    Customers c
LEFT JOIN
    Orders o ON c.id = o.customerId
WHERE
    o.id IS NULL;
```

### Question 8: Get the Nth Highest Salary

**Problem Statement:** Write a SQL query to get the `Nth` highest salary from the `Employees` table. For example, if `N=2`, return the second highest salary.

**Solution (using LIMIT/OFFSET for MySQL/PostgreSQL, where N is a variable):**
```sql
CREATE FUNCTION getNthHighestSalary(N INT) RETURNS INT
BEGIN
  SET N = N - 1;
  RETURN (
      SELECT DISTINCT salary
      FROM Employees
      ORDER BY salary DESC
      LIMIT 1 OFFSET N
  );
END
```

**Solution (using CTE and DENSE_RANK for SQL Server/Oracle, where N is a variable):**
```sql
CREATE FUNCTION getNthHighestSalary(N INT) RETURNS INT
AS
BEGIN
  RETURN (
    SELECT DISTINCT salary
    FROM (
      SELECT
        salary,
        DENSE_RANK() OVER (ORDER BY salary DESC) AS RankNum
      FROM Employees
    ) AS RankedSalaries
    WHERE RankNum = N
  );
END
```

### Question 9: Active Users in a given period

**Problem Statement:** Given a `Users` table (columns: `user_id`, `join_date`) and `Logins` table (columns: `login_id`, `user_id`, `login_date`). Find the number of active users in August 2023. An active user is defined as a user who has logged in at least once in that month.

**Solution:**
```sql
SELECT
    COUNT(DISTINCT l.user_id)
FROM
    Logins l
WHERE
    l.login_date >= '2023-08-01'
    AND l.login_date < '2023-09-01';
```

### Question 10: Departmental Salary Analysis

**Problem Statement:** For each department, find the average salary and the number of employees. Display department name, average salary, and employee count. Use `Employees` table (employee_id, employee_name, salary, department_id) and `Departments` table (department_id, department_name).

**Solution:**
```sql
SELECT
    d.department_name,
    AVG(e.salary) AS average_salary,
    COUNT(e.employee_id) AS employee_count
FROM
    Departments d
LEFT JOIN
    Employees e ON d.department_id = e.department_id
GROUP BY
    d.department_name
ORDER BY
    d.department_name;
```
